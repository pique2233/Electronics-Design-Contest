### 相机传感器
https://docs.singtown.com/micropython/zh/latest/openmvcam/library/omv.sensor.html

```python
sensor.reset()#初始化摄像头模块。

sensor.set_pixformat(sensor.RGB565)#设置图像的颜色格式为RGB565（16位颜色模式）。
'''sensor.GRAYSCALE：每像素8位。  通常我们使用RGB565
sensor.RGB565：每像素16位。
sensor.BAYER：每像素8位bayer模式。
sensor.YUV422：每像素16位（8位Y1，8位U，8位Y2，8位V，等等）
sensor.JPEG：压缩的 JPEG 数据。仅适用于 OV2640/OV5640。'''

sensor.set_windowing(120, 120)#裁剪图像，只取图像的中心部分80x80像素。
sensor.skip_frames(time=2000)#跳过初始帧，给摄像头一些时间来稳定。

sensor.set_framesize(sensor.QQVGA)#设置分辨率为QQVGA（160x120像素）。
# 设置相机模块的帧大小。 具体参考链接 
sensor.get_framesize()#->返回int

#sensor.set_brightness(brightness: int)→ None 设置相机图像亮度。-3 到 +3。
sensor.set_contrast(3)#设置图像对比度为3。 -3 到 +3。
sensor.set_saturation()# 饱和度 -3 到 +3
sensor.set_auto_gain(False)
sensor.set_auto_whitebal(False) # 这一步是设置白平衡  
# Flase 情况下可以设置RGB三个通道 (enable: bool, rgb_gain_db: Tuple[float, float, float] | None = None)→ None 
sensor.set_auto_exposure(False, 20000)#关闭自动增益、白平衡和曝光控制。


```
